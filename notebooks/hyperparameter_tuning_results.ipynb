```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd
```

```python
# Assuming you have an Optuna study object named `study`
# Extract trials from the study
trials = study.trials

# Create a DataFrame to store the trial information
data = pd.DataFrame({
    'width': [trial.params['width'] for trial in trials if 'width' in trial.params],
    'depth': [trial.params['depth'] for trial in trials if 'depth' in trial.params],
    'skip_block_layers': [trial.params['skip_block_layers'] for trial in trials if 'skip_block_layers' in trial.params],
    'objective_value': [trial.value for trial in trials]
})
```

```python

# Create 3D scatter plot
fig = go.Figure(data=[go.Scatter3d(
    x=data['width'],
    y=data['depth'],
    z=data['skip_block_layers'],
    mode='markers',
    marker=dict(
        size=5,
        color=data['objective_value'], # Color by objective value
        colorscale= 'Magma_r', #'Viridis', # Color scale. Can be any of "Cividis", "Inferno", "Magma", "Viridis", "Plasma". Add _r at the end for reversing
        colorbar=dict(title='Objective Value'),
        opacity=0.8
    )
)])

# Set titles for axes
fig.update_layout(scene=dict(
    xaxis_title='Width',
    yaxis_title='Depth (blocks)',
    zaxis_title='Skip Layers'
))


# Update layout with custom width and height
fig.update_layout(
    scene=dict(
        xaxis_title='Width',
        yaxis_title='Depth',
        zaxis_title='Skip Layers'
    ),
    width=1000,  # Increase width as needed
    height=800   # Increase height as needed
)

# Show the plot
fig.show()



```
