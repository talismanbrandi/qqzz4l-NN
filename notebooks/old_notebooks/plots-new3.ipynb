```python
import pandas as pd
import matplotlib.pyplot as plt
```

## h2 real and imaginary

```python
df_h2 = pd.read_csv('../models/skip-light-torch-4D-y5_y6-0503a9bc--16-32-leaky_relu-1024-adam-exp-schedule-mse-val_mse-96.535675-99.833029/test-results-0503a9bc-96.535675-99.833029.csv')
```

```python
df_h2
```

```python
df_h2.describe()
```

```python
plt.hist(df_h2[abs(df_h2.delta_y5) < 5].delta_y5, bins=100, histtype='step', label='h2 real', density=True)
plt.grid('dotted', alpha=0.5)
plt.xlabel(r'$\delta$')
plt.ylabel(r'pdf')
plt.legend()
plt.show()
```
